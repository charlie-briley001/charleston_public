# Binary Tree Level Order Traversal II
Given the root of a binary tree, return the bottom-up level order traversal of its nodes' values. (i.e., from left right, level by level from leaf to root).


Example 1:

Input: root = [3,9,20,null,null,15,7]

Output: [[15,7],[9,20],[3]]

```python
from collections import deque
from typing import List, Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def levelOrderBottom(self, root: Optional[TreeNode]) -> List[List[int]]:
        res = deque()
        if root is None:
            return []
        line = deque([root])

        while line:
            children = []
            for i in range(len(line)):
                node = line.popleft()
                children.append(node.val)
                if node.left:
                    line.append(node.left)
                if node.right:
                    line.append(node.right)
            res.appendleft(children)

        return list(res)
```

## Why this works
This is a breadth first problem, where each row or level of a tree needs to be visited and recorded prior to the next level.

Generally speaking BFS use a queue method where first nodes added are removed first. This problem takes on a similar format, but we use a deque method as it allows us to add more quickly to the front of a queue (which helps with our reverse order). We could have also done this with a list and just some indexing.

Algorithmically, the function first checks that the tree actually has a root, if not it returns None. If it does, the root becomes first in line / added to the queue. The challenge with this problem is being able to effectively captured every level together while then proceeding to the child nodes if applicable. How we get around this is by adding to the line of nodes to be checked. Since we start with the root, we have one node to check first. We first pop the node from the line since its being checked, we then grab its value, and store that in a list. We then check for any child nodes that need to be checked as well if they exist, we then add them to the line. So essentially we use the "for loop" to check a specific group of child nodes at a specific level, and use the "while" loop to proceed to the next level once ready.

 As we go through each level, we are capturing it completely within the "children" list which is then appended to the deque from the left side, hence the reverse ordering.
