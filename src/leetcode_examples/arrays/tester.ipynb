## Merge Sorted Array

You are given two integer arrays nums1 and nums2, sorted in non-decreasing order, and two integers m and n, representing the number of elements in nums1 and nums2 respectively.

Merge nums1 and nums2 into a single array sorted in non-decreasing order.

The final sorted array should not be returned by the function, but instead be stored inside the array nums1. To accommodate this, nums1 has a length of m + n, where the first m elements denote the elements that should be merged, and the last n elements are set to 0 and should be ignored. nums2 has a length of n.

```python
class Solution:
    def merge(self, nums1: list[int], m: int, nums2: list[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        del nums1[m:]
        if nums1 is None or nums2 is None:
            return nums1 or nums2
        idx = 0
        for n in nums2:
            while idx < len(nums1):
                if n < nums1[idx]:
                    break
                idx += 1
            nums1.insert(idx, n)
```

### Why this works

We start by removing the 0 placeholders as they actually complicate the problem more.

We then check if either list is None. If so, return the list that is not None.

We then move on to a pointer method on nums1, while looping over nums2. We loop over each element of nums2. Since nums2 and nums1 are already sorted we do not have to reset the pointer. We start by checking the first element of nums2 against index one of nums1. If nums2 < nums1 then insert nums2 value into that index, pushing all other elements of nums1 right. Again, since nums1 is in order already we know sequencing is preserved. We then move on to the next element of nums2 checking the same condition against the first index. If nums2 element is larger we iterate the index of nums1 and check the next, all the way until we are able to insert. If we get to the end of nums1 and cannot insert then we know that nums2 value is larger than all of nums1, and just insert into the end. Since nums2 is in order, we then can be certain remaining elements of nums2 would be appended. Hence, why we do not reset nums1 index.
