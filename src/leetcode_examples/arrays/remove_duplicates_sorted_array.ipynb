## Remove Duplicate Sorted Array

Given an integer array nums sorted in non-decreasing order, remove the duplicates in-place such that each unique element appears only once. The relative order of the elements should be kept the same.


```python
class Solution:
    def removeDuplicates(self, nums: list[int]):
        idx = 1
        while idx < len(nums):
            if nums[idx] in nums[:idx]:
                nums.pop(idx)
            else:
                idx += 1
        return len(nums)
```

### Why this works?

This is not the most efficient, but it is the first that came to my mind.

Essentially, since the first value of the array is always unique to everything before it (nothing), we started at index 1. We then check if the value at that index has occurred in the list subset that has occurred prior to it. If so pop that value from the array, and recheck the index with the newly adjust (in place) array. This will continue until it is a new, unique element, at which point we can iterate the index by one and check the next value. We do this until all values in the array are valid.
