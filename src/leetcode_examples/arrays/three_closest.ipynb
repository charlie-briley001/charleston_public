# 3Sum Closest
Given an integer array nums of length n and an integer target, find three integers at distinct indices in nums such that the sum is closest to target.

Return the sum of the three integers.

You may assume that each input would have exactly one solution.



```python
class Solution:
    def threeSumClosest(self, nums: list[int], target: int):
        nums.sort()
        closest = nums[0] + nums[1] + nums[2]
        for i in range(len(nums) - 2):
            left, right = i + 1, len(nums) - 1

            while left < right:
                sum1 = nums[i] + nums[left] + nums[right]

                if abs(target - sum1) < abs(target - closest):
                    closest = sum1

                if sum1 == target:
                    return target
                if sum1 < target:
                    left += 1
                else:
                    right -= 1

        return closest
```

### Why this works

We need to compare the sum of three values to a target value. To avoid randomly checking every combination of values, we use a pointer method by which we loop over each value once, and adjust two high and low pointers to search for the closest sum. We first sort the array.

We start by looping over each element of the array. With a left point at the following element and the right pointer at the last element of the array. We then check the sum difference from target against the previous closest value difference and update if it is closer. The reason we do this is if the difference is smaller than that of the then seen closest sum, then this group of numbers is closer to the target.

We then check a few more conditions. If the sum is target we return target because this means there are three values that sum to the target. In this instance there isn't any closer than we can get. Now if sum != target we must iterate left or right vale by one. If the current tested sum is < target then we can only get closer to the target by increasing the sum. Therefore we increment the left index by one. Since we have sorted the array this will result in a larger sum. If sum is greater than, we must decrease our current sum and therefore, increment right down by one. We continue this until right and left index cross, at which point we move on to the next value in the array.
