## Conver to Title

Given an integer columnNumber, return its corresponding column title as it appears in an Excel sheet.

```python
class Solution:
    def convertToTitle(self, columnNumber: int) -> str:
        letters = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
        high = ((columnNumber - 1) // 26)
        low = ((columnNumber - 1) % 26)

        if high == 0:
            return letters[low]

        return self.convertToTitle(high) + letters[low]
```

### Why this works

We have listed our the ABCs across index 0-25. To then properly map to an integer it simply becomes a challenge of trying to identify remainders and their index.

We start by finding how many times 26 goes into the column number evenly (high value). We then also want to find the remainder (low value). The high value will tell us how many iterations through A-Z we must go through, while the low value tells us which letter in the final iteration (right most letter in the string response). Since we can capture the right most index through this low value we just need to append to the left string values. We get this by simply treating the high value as its own columnNumber, and find the string that it is associated with. This process will continue until the high value is zero.
