## Valid Palindrome

A phrase is a palindrome if, after converting all uppercase letters into lowercase letters and removing all non-alphanumeric characters, it reads the same forward and backward. Alphanumeric characters include letters and numbers.

Given a string s, return true if it is a palindrome, or false otherwise.

```
class Solution:
    def isPalindrome(self, s: str) -> bool:

        letters = [
            l.lower()
            for l in s
            if l.isalnum()
            ]

        return letters == letters[::-1]
```

### Why this works

The function takes in a string, which may have both alphanumeric or non - alphanumeric characters. With regard to this problem it doesn't matter, we must remove non - alphanumeric, hence why we use list comp to create a cleaned list of letters.

Then to check for a valid palindrome, I simply reverse the list using index slicing and compare the two lists. If they are equal, the initial string is a valid palindrome.
