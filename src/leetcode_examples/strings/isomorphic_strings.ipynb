## Isomorphic Strings

Given two strings s and t, determine if they are isomorphic.

Two strings s and t are isomorphic if the characters in s can be replaced to get t.

All occurrences of a character must be replaced with another character while preserving the order of characters. No two characters may map to the same character, but a character may map to itself.

```python
class Solution:
    def isIsomorphic(self, s: str, t: str) -> bool:
        s_mapping = {}
        t_mapping = {}
        for i in range(len(s)): # s and t will always equal so can make this assumption
            if s[i] not in s_mapping:
                s_mapping[s[i]] = i

            if t[i] not in t_mapping:
                t_mapping[t[i]] = i

            if s_mapping[s[i]] != t_mapping[t[i]]:
                return False
        return True
```

### Why this works

Essentially this algorithm is based on the fact that two characters must be in the same index constantly for the strings to be isomorphic.

To do this we initial two dictionaries to track each character and their first index value seen. Then we loop over each character in the strings t and s at the same time. Note that since the problem states length of t is equal to s, we can use this approach. We then check if the character value is already in the mappings. If not then add, otherwise skip to the conditional where we ensure the index value matches for both characters in t and s.

This works because if in the same loop both characters from t and s are in the mapping already then they must have the same index to be isomorphic. We don't update the index if a character pair is seen later on in the string rather than the first time, because it would be redundant - both indices would match. This is why we only update characters that do not exist in the mapping. If one character is added but the other is not, then that means you have disjoint index pairs, which is caught by the conditional.
